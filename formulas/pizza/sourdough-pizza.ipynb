## Sourdough pizza

from https://www.thefreshloaf.com/node/34452/pizza-bliss

```python
import sys
sys.path.append('../..')
from breadsheet_utils import RecipeCalculator, create_formula, format_and_display
```

```python
calc = RecipeCalculator(
    num_loaves=4, 
    #weight_grams=296.5,
    weight_grams=300,
    waste_factor=0
)
calc.print_batch_info()
```

```python
pre_fermented_flour_ratio=.1
title='Sourdough Pizza'

formula = create_formula({
    "bread-flour": 100,
    "water": 70, 
    "salt": 2
})

levain = create_formula({
    "bread-flour": 100,
    "water": 100,
    "seed": 20
})

steps='''
- bulk ferment 5-14 hours, fold every 30 min 2-4 times
- divide & shape
- refrigerate 48-72 hours
'''

levain_weights, final_dough = calc.calculate(formula, levain_df=levain, pre_fermented_flour_ratio=pre_fermented_flour_ratio)
format_and_display(formula=final_dough, levain=levain_weights, calc=calc, title=title, steps=steps, reserved_seed_grams=150)
```
