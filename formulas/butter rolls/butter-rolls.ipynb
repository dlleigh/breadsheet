## Butter rolls

```python
import sys
sys.path.append('../..')
from breadsheet_utils import RecipeCalculator, create_formula, format_and_display
```

```python
pre_fermented_flour_ratio = .25

formula = create_formula({
    "bread-flour": 100,
    "water": 47,
    "egg": 10,
    "brown butter": 14,
    "sugar": 6,
    "milk powder": 5,
    "salt": 2,
    "yeast": 0.8, # 1.0 or 0.8 for overnight retard
})

poolish = create_formula({
    "bread-flour": 100,
    "water": 100, 
    "yeast": 0.088
})

steps= '''
- bulk ferment 1 hr
- retard overnight
- shape
- proof
- brush with melted butter (or top with grated parm)
- bake at 400
- brush again with melted butter
'''
title="Butter rolls"
```

```python
calc = RecipeCalculator(
    num_loaves=[24,12], 
    weight_grams=80,
    preferment_waste_factor=0.04,
    waste_factor=0.02,
)
calc.print_batch_info()
```

```python
poolish_weights, final_dough = calc.calculate(formula, poolish_df=poolish, 
                                              pre_fermented_flour_ratio=pre_fermented_flour_ratio)

format_and_display(formula=final_dough, poolish=poolish_weights, calc=calc, title=title, steps=steps)
```
