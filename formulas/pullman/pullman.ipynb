## Pullman

```python
import sys
sys.path.append('../..')
from breadsheet_utils import RecipeCalculator, create_formula, format_and_display
```

```python
formula = create_formula({
    "bread-flour": 100,
    "water": 60,
    "butter": 5,
    "sugar": 2.5,
    "milk powder": 5,
    "salt": 1.8,
    "yeast": .83,
})
steps= '''
- desired dough temp 78-80
- bulk ferment 2hrs, fold once
- preshape round, rest, then shape into cylinder with no taper
- oil pans and lids
- pans should be half full
- proof 60-90 min with lids
- bake at 400 40-45min
'''
title="Pullman"
```

```python
# using 13 x 3.75 x 3.75 pans with lids

calc = RecipeCalculator(
    num_loaves=1, 
    weight_pounds=2.3,
    waste_factor=0.02,
)
calc.print_batch_info()
```

```python
final_dough = calc.calculate_straight_dough(formula)
format_and_display(final_dough, calc, title=title, steps=steps)
```
